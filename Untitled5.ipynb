```python
# a) engagement_score - MUST
data['engagement_score'] = data['likes'] + data['comments']*2 + data['shares']*3

# b) log-transformed metrics - MUST
import numpy as np
data['log_likes'] = np.log1p(data['likes'])
data['log_impressions'] = np.log1p(data['impression_count'])
data['log_followers'] = np.log1p(data['follower_count'])

# c) hashtag_count - MUST
data['hashtag_count'] = data['hashtags'].astype(str).apply(lambda x: x.count('#'))

print(data[['likes','engagement_score','log_likes','hashtag_count']].head())
```

```python
# Full stats one shot
cols = ['likes','comments','shares','watch_time_sec','engagement_rate','follower_count']

# 1. Mean, Median, Mode
print("--- Mean, Median, Mode ---")
for col in cols:
    mean_val = data[col].mean()
    median_val = data[col].median()
    mode_val = data[col].mode()[0]
    print(f"{col} -> Mean: {mean_val:.2f}, Median: {median_val:.2f}, Mode: {mode_val}")

# 2. Standard deviation, Variance
print("\n--- Std Dev & Variance ---")
print(data[cols].std())
print(data[cols].var())

# 3. Percentiles
print("\n--- Percentiles (25%, 50%, 75%, 95%) ---")
print(data[cols].quantile([0.25, 0.5, 0.75, 0.95]))

# 4. Skewness and Kurtosis (Optional but full marks kittan cheyyam)
print("\n--- Skewness & Kurtosis ---")
print("Skewness:")
print(data[cols].skew())
print("\nKurtosis:")
print(data[cols].kurtosis())

# 5. Full describe in one line
print("\n--- Full Describe ---")
print(data[cols].describe().T) # T = transpose, better view
```

```python

text_col = None
for col in data.columns:
    if 'content' in col.lower() or 'text' in col.lower() or 'post' in col.lower() or 'message' in col.lower() or 'ticket' in col.lower():
        text_col = col
        break

if text_col is None:
    text_col = data.select_dtypes(include='object').columns[0] # first text column

print(f"Text column found: {text_col}")


data['clean_text'] = data[text_col].astype(str).str.lower()
data['clean_text'] = data['clean_text'].str.replace(r'[^\w\s]', '', regex=True)
data['clean_text'] = data['clean_text'].str.replace(r'\s+', ' ', regex=True).str.strip()


slang_dict = {'u':'you','ur':'your','gr8':'great','b4':'before'}
def fix_slang(t):
    for k,v in slang_dict.items():
        t = t.replace(f" {k} ", f" {v} ")
    return t
data['clean_text'] = data['clean_text'].apply(fix_slang)

print("SUCCESS - Cleaning Done!")
print(data[['clean_text']].head(2))
```

```python

post_list = data[text_col].tolist()[:5]
print("List example:", post_list)


word_counts_dict = data['clean_text'].str.split().explode().value_counts().head(5).to_dict()
print("\nDictionary (Top words):", word_counts_dict)
```

```python

def count_words(text):
    return len(str(text).split())
data['word_count'] = data['clean_text'].apply(count_words)


high = 0
for wc in data['word_count']:
    if wc > 20:
        high += 1
print(f"Long posts >20 words: {high}")

def label_len(n):
    if n > 30: return 'Long'
    elif n > 15: return 'Medium'
    else: return 'Short'

data['length_label'] = data['word_count'].apply(label_len)
print(data['length_label'].value_counts())
```

```python

import pandas as pd

# Auto find numeric column for rating
num_cols = data.select_dtypes(include='number').columns.tolist()
rating_col = num_cols[0] # first numeric
text_cols = data.select_dtypes(include='object').columns.tolist()
text_col = text_cols[0] if text_cols else data.columns[0]

avg_rating = data[rating_col].mean()
print(f"Average Rating ({rating_col}): {avg_rating:.2f}")


data['len'] = data[text_col].astype(str).apply(len)
longest_row = data.loc[data['len'].idxmax()]
print(f"\nLongest Feedback Length: {longest_row['len']}")
print(f"Longest Feedback Text: {str(longest_row[text_col])[:200]}...")


data['clean'] = data[text_col].astype(str).str.lower().str.replace(r'[^\w\s]', '', regex=True)
all_words = ' '.join(data['clean']).split()
print(f"\nTotal Unique Words: {len(set(all_words))}")
```

```python

sorted_data = data.sort_values(by=rating_col, ascending=False).head(5)

ratings = sorted_data[rating_col].tolist()
feedbacks = sorted_data[text_col].astype(str).str[:50].tolist()

print("Top 5 Ranking:")
for i in range(len(ratings)):
    print(f"{i+1}. Rating={ratings[i]} | Text={feedbacks[i]}")


my_dict = dict(zip(ratings, feedbacks))
print("\nStructured Dict (using zip):")
print(my_dict)
```

```python

print("---- SUMMARY TABLE ----")
summary_df = pd.DataFrame({
    "Metric": ["Average Rating", "Longest Feedback", "Total Tickets", "Unique Words"],
    "Value": [f"{avg_rating:.2f}", f"{data['len'].max()} chars", len(data), len(set(all_words))]
})
print(summary_df.to_string(index=False))

print("\n---- Top Words ----")
top_words = pd.Series(all_words).value_counts().head(10)
print(top_words.to_string())
```
